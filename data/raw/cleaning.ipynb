```python
df.rename(columns={"OrderDate":"order_date"},inplace=True)
df.rename(columns={"CustomerName":"customer_name"},inplace=True)
```

```python
df["Price"]=df["Price"].astype(float)
```

```python
df["Quantity"]=df["Quantity"].astype(int)
```

```python
df["total_amount"]=df["Price"]*df["Quantity"]
```

KPI analysis

```python
Total_revenue=df["total_amount"].sum()
print(Total_revenue)
```

```python
avrage_order_value=df.groupby("OrderID")["total_amount"].sum().mean()
print(avrage_order_value,"avrage_order_value")
```

```python
hsp=df.groupby("Product")["Quantity"].sum().sort_values(ascending=False).head(1)
print(hsp)
```

```python
mcc=df.groupby("City")["Quantity"].count().sort_values(ascending=False).head(1)
print(mcc)
```
